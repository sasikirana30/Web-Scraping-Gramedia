### Import Library , Membuat Web driver , Menentukan url dan melakukan scrolling halaman
    

```python
# data manipulation or dataframe
import time
import pandas as pd

# webscraping
from selenium import webdriver
from bs4 import BeautifulSoup 

# create driver
driver = webdriver.Chrome() 

# link
url = 'https://www.gramedia.com/search?q=novel'

# access the website
driver.get(url) 

# Scroll beberapa kali agar lebih banyak produk muncul 

for i in range(10):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    
```

### Mengambil Source HTML dan Membuat Objek BeautifulSoup

```python
# fetch html components
html = driver.page_source 

# parse
soup = BeautifulSoup(html, "html.parser")
```

```python
print(soup.prettify()[:5000])
```

#### List HTML yang diambil dari web gramedia 

```python
#HTML

judul : 
<h2 class="relative line-clamp-2 text-xs-medium text-neutral-700" title="Novel The Big Four (Empat Besar)" data-testid="productCardTitle">Novel The Big Four (Empat Besar)</h2>
<h2 class="relative line-clamp-2 text-xs-medium text-neutral-700" title="Novel Cantik Itu Luka" data-testid="productCardTitle">Novel Cantik Itu Luka</h2>
<h2 class="relative line-clamp-2 text-xs-medium text-neutral-700" title="Five Nights At Freddy`S Graphic Novel #1: The Silver Eyes" data-testid="productCardTitle">Five Nights At Freddy`S Graphic Novel #1: The Silver Eyes</h2>

Penulis: 
<div class="relative truncate text-2xs-medium text-neutral-500" title="Agatha Christie" data-testid="productCardAuthor">Agatha Christie</div>
<div class="relative truncate text-2xs-medium text-neutral-500" title="Eka Kurniawan" data-testid="productCardAuthor">Eka Kurniawan</div>

Harga:
<div class="text-s-extrabold text-neutral-700" data-testid="productCardFinalPrice">Rp51.750</div>
<div class="text-s-extrabold text-neutral-700" data-testid="productCardFinalPrice">Rp100.000</div>

Diskon: 
<div class="ProductCard_discount__Q8rno inline-block rounded-2xs bg-red-50 px-1 text-2xs-medium text-red-500" data-testid="productCardDiscount">10%</div>
<div class="ProductCard_discount__Q8rno inline-block rounded-2xs bg-red-50 px-1 text-2xs-medium text-red-500" data-testid="productCardDiscount">15%</div>
```

### Melakukan Looping Data

Untuk mengambil data Judul buku , harga , penulis dan diskon 

```python
# empty list

titles = []
prices = []
authors = []
Discount = []

# Judul

el_titles = soup.find_all('h2', {'class': 'relative line-clamp-2 text-xs-medium text-neutral-700'}) 
for i in el_titles:
    titles.append(i.get_text())

# Harga

el_prices = soup.find_all('div', {'class': 'text-s-extrabold text-neutral-700'}) 
for i in el_prices:
    prices.append(i.get_text())

# Penulis
    
el_authors = soup.find_all('div', {'class': 'relative truncate text-2xs-medium text-neutral-500'})
for i in el_authors:
    authors.append(i.get_text())
    
# Diskon

el_Discount = soup.find_all('div',{'class': 'ProductCard_discount__Q8rno inline-block rounded-2xs bg-red-50 px-1 text-2xs-medium text-red-500'})
for i in el_Discount:
    Discount.append(i.get_text()) 

```

### Validasi dan Verifikasi Konsistensi Data

```python
print("judul   :", len(titles))
print("harga   :", len(prices))
print("penulis :", len(authors))
print("diskon  :", len(Discount))
```

### Catatan : 

Setelah dilakukan pengecekan panjang list menggunakan funngsi 'len()' terdapat perbedaan pada jumlah data "diskon" dibandingkan yang lainnya. Penyebabnya tidak semua produk buku dihalaman website gramedia memiliki program diskon. 

### Mengambil seluruh data produk menggunakan find_all()


```python
boxes = soup.find_all('div', {'data-testid': 'productCardContent'}) 
```

### Melakukan Looping Data 

```python
# empty list

titles = []
prices = []
authors = []
Discount = []

for box in boxes: 
    # Judul 
    try : 
        title = box.find('h2', {'class': 'relative line-clamp-2 text-xs-medium text-neutral-700'}) 
        title = title.get_text(strip=True) if title else None
    except: 
        title = None
    titles.append(title)

    # Harga
    try :
        price = box.find('div', {'class': 'text-s-extrabold text-neutral-700'})
        price = price.get_text(strip=True) if price else None
    except:
        price = None
    prices.append(price) 

    # Penulis
    try:
        author = box.find('div', {'class': 'relative truncate text-2xs-medium text-neutral-500'})
        author = author.get_text(strip=True) if author else None
    except:
        author = None
    authors.append(author)

    # Diskon
    try:
        discount = box.find('div',{'class': 'ProductCard_discount__Q8rno inline-block rounded-2xs bg-red-50 px-1 text-2xs-medium text-red-500'})
        discount = discount.get_text(strip=True) if discount else None
    except:
        discount = None
    Discount.append(discount) 
```

### Menyimpan Hasil Scraping ke DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Judul": titles,
    "Harga": prices,
    "Penulis": authors,
    "Diskon": Discount
})
```

### Pembersihan dan Transformasi Kolom Diskon

Pada tahap ini dilakukan pembersihan data pada kolom `Diskon`. Simbol `%` dihapus, nilai `NaN` diganti menjadi `0`, kemudian seluruh data dikonversi menjadi tipe numerik (`integer`) agar dapat digunakan untuk analisis dan penyimpanan ke database.

```python
df['Diskon'] = (
    df['Diskon']
    .astype(str)
    .str.replace('%', '', regex=False)
    .replace('None', '0')
)

df['Diskon'] = pd.to_numeric(df['Diskon'], errors='coerce').fillna(0).astype(int)
```

### Pembersihan dan Transformasi Kolom Harga

Pada tahap ini dilakukan pembersihan data pada kolom `Harga`. Karakter non-numerik seperti simbol `Rp` dan tanda titik (`.`) sebagai pemisah ribuan dihapus. Setelah itu, data dikonversi menjadi tipe numerik agar dapat digunakan untuk analisis dan penyimpanan ke database PostgreSQL.

```python
df['Harga'] = (
    df['Harga']
    .str.replace('Rp', '', regex=False)
    .str.replace('.', '', regex=False)  
    .str.strip()
)

df['Harga'] = pd.to_numeric(df['Harga'], errors='coerce')
```
